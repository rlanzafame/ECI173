# Stress Plot

Create a plot of stress with depth for various surface load conditions.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# define functions that will be used to plot specific lines
def stress_model_1(z, q):
    """Calculate stress with depth for a surface load q using a simple linear model."""
    return q * np.exp(-z / 10)

def stress_model_2(z, q):
    """Calculate stress with depth for a surface load q using athe 2:1 method."""
    return q / (1 + z / 10)

```

```python
# Create z values (depth)
z = np.linspace(0, 50, 100)
q = 100  # Surface load

# Create figure and axis
fig, ax = plt.subplots(figsize=(8, 6))

# Plot stress models
stress_1 = stress_model_1(z, q)
stress_2 = stress_model_2(z, q)

ax.plot(stress_1, z, label='Model 1 (Exponential)', linewidth=2)
ax.plot(stress_2, z, label='Model 2 (2:1 Method)', linewidth=2)

# Invert y-axis so depth increases downward
ax.invert_yaxis()
ax.xaxis.tick_top()
ax.xaxis.set_label_position('top')
# Labels and legend
ax.set_xlabel('Stress', fontsize=12)
ax.set_ylabel('Depth (z)', fontsize=12)
ax.set_title('Stress with Depth for Surface Load Conditions', fontsize=14)
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
