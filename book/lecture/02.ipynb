# Stress Plot

Create a plot of stress with depth with different elastic models for various surface load conditions.

_GitHub Copilot was used to help write this code by prompting it with suggestions for function and figure layouts, then each model was defined by the author._ 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def stress_model_1(z, B=1, L=1):
    """Model 1."""

    z = np.where(np.asarray(z) == 0, 1e-6, z)

    m = B/z/2
    n = L/z/2
    C1 = 1 + m**2 + n**2
    C2 = (m*n)**2

    if sum(C2>C1)>0:
        print(f"C2 is greater than C1 in {sum(C2>C1)} out of {len(C1)} points.")

    angle = np.arctan(m*n*np.sqrt(C1)/(C1 - C2))
    angle[(C2 > C1) & (angle < 0)] += np.pi

    return 1/np.pi*(2*m*n*np.sqrt(C1)/(C1 + C2)*(1 + C1)/C1
                    + angle)

def stress_model_2(z, B=1, L=1):
    """Model 2."""
    return 1/(1 + z/B)/(1 + z/L)

def stress_model_3(z, B=1, L=1):
    """Model 3."""

    z = np.where(np.asarray(z) == 0, 1e-6, z)

    theta = np.arctan(B/2/z)

    return 1/2/np.pi*(4*theta + 2*np.sin(2*theta))

def stress_model_4(z, B=1, L=1):
    """Model 4."""

    R1 = np.sqrt((B/2)**2 + z**2)
    R2 = np.sqrt((L/2)**2 + z**2)
    Rs = np.sqrt((B/2)**2 + (L/2)**2 + z**2)

    term1 = np.arctan(B/2*L/2/z/Rs)
    term2 = B/2*L/2*z/Rs*(1/R1**2 + 1/R2**2)

    return 4/2/np.pi*(term1 + term2)
```

```python
z = np.linspace(1e-6, 5, 1000)
width = 1 
length = 200000

fig, ax = plt.subplots(figsize=(8, 6))

stress_1 = stress_model_1(z, width, length)
stress_2 = stress_model_2(z, width, length)
stress_3 = stress_model_3(z, width, length)
stress_4 = stress_model_4(z, width, length)

my_linewidth = 3
ax.plot(stress_1, z, label='Model 1', linewidth=my_linewidth)
ax.plot(stress_2, z, label='Model 2', linewidth=my_linewidth)
ax.plot(stress_3, z, label='Model 3', linewidth=my_linewidth)
ax.plot(stress_4, z, label='Model 4', linewidth=my_linewidth, linestyle='--')

ax.invert_yaxis()
ax.xaxis.tick_top()
ax.xaxis.set_label_position('top')

ax.set_xlabel('Stress', fontsize=12)
ax.set_ylabel('Depth (z)', fontsize=12)
ax.set_title('Stress with Depth for Surface Load Conditions', fontsize=14)
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
